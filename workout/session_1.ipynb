### Why LinearRegression gave wrong predictions
A plain LinearRegression fits a model of the form y = a * x + b (or y = a1*x1 + a2*x2 + ... + b for multiple features).
In your original dataset you attempted to predict `cubs` (nums**3) from `nums` alone using a linear model. A cubic relationship cannot be represented exactly by a single linear feature — the model will try to approximate the nonlinear curve with a straight line and will give very poor predictions, especially outside the training range or when you trained using different features.
Fixes: 1) Use PolynomialFeatures to add x^2 and x^3 terms and then fit LinearRegression. 2) Use a nonlinear model (DecisionTree/RandomForest/NeuralNet). 3) If the target is exactly the cube, simply compute nums**3 directly. Below are two practical cells showing (A) polynomial regression degree=3 and (B) direct computation.

```python
# Polynomial regression example: fit nums -> cubs using x, x^2, x^3
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# prepare data (nums and cubs from existing df)
X_poly = df[["nums"]].copy()
y_poly = df["cubs"].copy()

# generate polynomial features degree 3
poly = PolynomialFeatures(degree=3, include_bias=False)
Xp = poly.fit_transform(X_poly)

# fit linear regression on polynomial features
poly_model = LinearRegression()
poly_model.fit(Xp, y_poly)

# predictions on training set and new points
pred_train = poly_model.predict(Xp)
print('R^2 on training set:', r2_score(y_poly, pred_train))

new_df = pd.DataFrame({
    "nums": [5, 10, 15]
})
new_Xp = poly.transform(new_df[["nums"]])
print('Polynomial model predictions:', poly_model.predict(new_Xp))

# Show polynomial feature names for clarity
try:
    print('Feature names:', poly.get_feature_names_out(['nums']))
except Exception:
    pass
```
