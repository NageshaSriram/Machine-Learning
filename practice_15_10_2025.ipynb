```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
X = np.array([12, 9, 3, 6, 24, 13, 21, 6, 16, 13, 54, 23, 46, 32, 87, 23, 34])
sns.boxplot(X)
```

```python
sns.histplot(X, kde=True, bins=5, color='skyblue')
plt.title('Histogram with KDE')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

```python
import pandas as pd 
df = pd.DataFrame(X, columns=['Values'])
df
```

```python
df['Values'].skew()
```

```python
df['Values'] = np.log1p(df['Values'])  # log(1 + x) to handle zero values
df['Values'].skew()
```

```python
sns.histplot(df['Values'], kde=True, bins=5, color='skyblue')
plt.title('Histogram with KDE')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

```python
# Plot histogram (bars) and KDE (line) to check normal distribution
sns.histplot(X, kde=True, stat="density", bins=10, color="skyblue", edgecolor="black")
plt.title("Histogram and KDE Plot")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```
